# plots of gaze and fixations on screen

```python
import pandas as pd
df_pos = pd.read_csv("data/sentence_position.csv")
```

```python
presentation = 2

results = gaze2.gaze_on_screen(df_pos)

gaze2.plot_gaze_on_text(df_pos, pres=presentation, results=results)

sacc = gaze2.saccades_movement(pres=presentation, df_position=df_pos)
sacc["is_regression"].mean()          # overall regression rate
sacc["saccade_type"].value_counts()   # breakdown: line vs within-line regressions

gaze2.plot_saccades_on_text(df_pos, pres=presentation, saccades=sacc, show_gaze=True, results=results)
```

```python
reggs = 0
pres_regs = []
for i in range(96):
    fixations = gaze2.filter_part2_to_part1_saccades(pres=i, df_position=df_pos)
    if len(fixations) > 1:
        reggs += len(fixations)
        pres_regs.append(i)
        # print(f"Presentation {i}: {len(fixations)} fixations on part 2 that were preceded by a fixation on part 1.")
print(f"Total regressions: {pres_regs} presentations with regressions, {reggs} total regressions.")
```
